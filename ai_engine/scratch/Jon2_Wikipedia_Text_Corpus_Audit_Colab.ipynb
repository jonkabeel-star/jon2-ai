# Jon2 AI — Audit the 130k Arabic Wikipedia text corpus\nThis notebook locates `wikipedia_ar_cc_by_sa.txt` in Google Drive or the current Colab runtime and reports aggregate statistics only. It never prints or uploads the source text.\n\n**Important:** the historical file was reported as about 130,002 characters from 52 Arabic Wikipedia articles. This notebook re-checks the actual file; it does not assume the old count is still present. CC BY-SA 4.0 does not automatically settle every question about model training or commercial use. Verify each page's license, attribution and share-alike requirements before including it.

```python
from google.colab import drive\ndrive.mount('/content/drive')\nfrom pathlib import Path\nsearch_roots = [Path('/content/drive/MyDrive'), Path('/content/jon2-ai/ai_engine/training_data')]\nmatches = []\nfor root in search_roots:\n    if root.exists():\n        matches.extend(root.rglob('wikipedia_ar_cc_by_sa.txt'))\nmatches = sorted(set(matches))\nprint('Matching files:', len(matches))\nfor p in matches[:10]: print(f'{p} — {p.stat().st_size:,} bytes')\nif len(matches) != 1:\n    raise RuntimeError('Expected exactly one corpus file. If the old file is only in a previous temporary Colab session, upload/copy it into Google Drive first. No text was displayed.')\nCORPUS_FILE = matches[0]\n
```

```python
import urllib.request\naudit_url = 'https://raw.githubusercontent.com/jonkabeel-star/jon2-ai/feature/jon2-from-scratch/ai_engine/scratch/audit_text_corpus.py'\naudit_script = Path('/content/audit_text_corpus.py')\nurllib.request.urlretrieve(audit_url, audit_script)\nprint('Downloaded aggregate-only audit utility.')\n
```

```python
import json, subprocess, sys\nresult = subprocess.run([sys.executable, str(audit_script), '--file', str(CORPUS_FILE)], capture_output=True, text=True)\nprint(result.stdout)\nif result.returncode != 0: raise RuntimeError('Text corpus audit did not pass; inspect the aggregate report above.')\n
```
