```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
roots = [Path('/content/drive/MyDrive'), Path('/content/jon2-ai/ai_engine/training_data')]
matches = sorted({p for root in roots if root.exists() for p in root.rglob('wikipedia_ar_cc_by_sa.txt')})
print('Matching files:', len(matches))
for p in matches[:10]:
    print(f'{p} — {p.stat().st_size:,} bytes')
if len(matches) != 1:
    raise RuntimeError('Expected exactly one corpus file. Copy the file into Google Drive first if needed.')
CORPUS_FILE = matches[0]
```

```python
import urllib.request, subprocess, sys
audit_url = 'https://raw.githubusercontent.com/jonkabeel-star/jon2-ai/feature/jon2-from-scratch/ai_engine/scratch/audit_text_corpus.py'
audit_script = Path('/content/audit_text_corpus.py')
urllib.request.urlretrieve(audit_url, audit_script)
result = subprocess.run([sys.executable, str(audit_script), '--file', str(CORPUS_FILE)], capture_output=True, text=True)
print(result.stdout)
if result.returncode != 0:
    raise RuntimeError('Audit did not pass; review the aggregate report above.')
```
